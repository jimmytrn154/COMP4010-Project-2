# Monthly Combined Risk Score Training

This notebook trains and benchmarks forecasting models for the monthly combined risk score target used by the Mekong Lens dashboard.

## Feature engineering plan

- autoregressive lags: current value plus 1, 2, 3, 6, 12 month lags
- rolling stability features: rolling mean, std, min, max over 3, 6, 12 months
- seasonality: forecast-month sine/cosine, quarter, time index
- province encoding via one-hot encoding inside the preprocessing pipeline
- historical priors: province-level and province-month target averages
- known context features: monthly rainfall climatology, water climatology, exposure proxies
